```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({
    'marks':[80,70,90,60,90,None],
    'attendence':[75,80,90,67,80,None],
    'department':['CSE','CSE','CSE','CSE','ECE',None]
})
#mean imputation
df['marks']=df['marks'].fillna(df['marks'].mean())
#median imputation
df['attendence']=df['attendence'].fillna(df['attendence'].median())
#forward fill(fills none with previous or above values)
df['attendence']=df["attendence"].ffill()
#Fill missing categorical value using mode
df['department']=df['department'].fillna(df['department'].mode()[0])
#Label Encoding
encoder=LabelEncoder()
df["department"]=encoder.fit_transform(df['department'])
print(df)
```

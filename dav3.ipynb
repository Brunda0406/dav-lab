```python
import pandas as pd
import matplotlib.pyplot as pit
from google.colab import files
uploaded=files.upload()

```

```python
df=pd.read_csv(r"diabetes_dataset.csv")
df.head()
```

```python
df.columns
```

```python
pit.hist(df['Glucose'],bins=10)
pit.xlabel('Glucose Level')
pit.ylabel('Frequency')
pit.title('Distribution of Glucose Levels')
pit.show()
```

```python
pit.boxplot(df['Glucose'])
pit.ylabel('Glucose Level')
pit.title('Boxplot of Glucose Levels')
pit.show()
```

```python
pit.scatter(df['Age'],df['Glucose'])
pit.xlabel('Age')
pit.ylabel('Glucose Level')
pit.title('Glucose Level vs Age')
pit.show()
```

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Array:", a)
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
print("Shape:", a.shape)
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Asha"],
    "Marks": [80, 90, 75]
}

df = pd.DataFrame(data)

print("Data:")
print(df)

print("First row:")
print(df.head(1))

print("Marks column:")
print(df["Marks"])

print("Total marks:", df["Marks"].sum())
print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())

print("Sorted data:")
print(df.sort_values("Marks"))
```
